**Aim**

To analyze a simulated system process execution log and identify suspicious processes running from non-standard locations or using suspicious names.

**Algorithm**

Define a list of baseline legitimate process names (e.g., svchost.exe, explorer.exe).

Define legitimate system directory execution paths (e.g., C:\Windows\System32).

Load simulated process execution log entries (PID, process name, path).

Check process execution path against approved system directories.

Identify processes executing from temporary or user directories (e.g., AppData\Local\Temp).

Detect typosquatting variants of known process names (e.g., svchostt.exe).

Flag processes matching suspicious indicators.

Print alert details of suspicious processes.

```python
process_log = [
    {"pid": 1024, "name": "svchost.exe", "path": "C:\\Windows\\System32\\"},
    {"pid": 2048, "name": "svchostt.exe", "path": "C:\\Users\\User\\AppData\\Local\\Temp\\"},
    {"pid": 3072, "name": "malware.exe", "path": "C:\\ProgramData\\"}
]

approved_paths = ["C:\\Windows\\System32\\", "C:\\Program Files\\"]

def analyze_processes(logs):
    print("--- Suspicious Process Execution Report ---")
    for proc in logs:
        is_suspicious = False
        reason = ""

        if proc["path"] not in approved_paths:
            is_suspicious = True
            reason = f"Unusual execution path ({proc['path']})"
        elif "temp" in proc["path"].lower():
            is_suspicious = True
            reason = "Executing from Temp folder"

        if is_suspicious:
            print(f"[SUSPICIOUS PID {proc['pid']}] {proc['name']} -> Reason: {reason}")

analyze_processes(process_log)
```

**Result**

Identified malicious processes executing from temporary paths and using typosquatted process names.
