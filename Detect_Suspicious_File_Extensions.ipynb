**Aim**

To analyze filenames in a folder and flag suspicious extension patterns, such as double extensions (invoice.pdf.exe).

**Algorithm**

Import the os module for directory iteration.

Define lists of dangerous extensions (e.g., .exe, .vbs, .bat, .ps1).

Set target directory path to inspect.

List all filenames in the target directory.

Split filenames using dot . delimiters.

Check if a filename contains more than one extension suffix.

Compare trailing extension against known dangerous execution types.

Flag double-extension files and suspicious executable disguises.

```python
import os

sample_files = ["document.pdf", "invoice.pdf.exe", "report.docx", "data.csv.vbs", "image.png"]
dangerous_exts = [".exe", ".vbs", ".bat", ".scr", ".ps1"]

def detect_suspicious_extensions(file_list):
    for filename in file_list:
        parts = filename.split(".")
        if len(parts) > 2 and f".{parts[-1]}" in dangerous_exts:
            print(f"[ALERT] Double extension detected: {filename}")
        elif f".{parts[-1]}" in dangerous_exts:
            print(f"[WARNING] Executable file found: {filename}")

detect_suspicious_extensions(sample_files)
```

**Result**

Flagged suspicious double-extension files masking executable payloads under benign document names.
