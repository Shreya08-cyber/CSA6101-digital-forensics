**Aim**

To detect potential ransomware activity based on rapid file modification rates and sudden abnormal extension changes.

**Algorithm**

Define activity log entries containing file modification events and timestamps.

Group modification events by time window (e.g., modifications per minute).

Set rate threshold for rapid modifications (e.g., > 5 files/min).

Define known ransomware extensions list (e.g., .locked, .crypto, .enc).

Scan modification logs for sudden changes in file extensions.

Calculate file modification frequency within active time windows.

Flag activity exceeding threshold as potential ransomware outbreak.

Output security alert with affected files and process metrics.

```python
from collections import Counter

file_logs = [
    {"time": "10:00:01", "file": "doc1.docx.locked", "ext": ".locked"},
    {"time": "10:00:02", "file": "doc2.xlsx.locked", "ext": ".locked"},
    {"time": "10:00:03", "file": "doc3.pptx.locked", "ext": ".locked"},
    {"time": "10:00:04", "file": "doc4.pdf.locked", "ext": ".locked"},
    {"time": "10:00:05", "file": "doc5.png.locked", "ext": ".locked"}
]

ransom_exts = [".locked", ".crypto", ".enc"]

def detect_ransomware_behavior(logs, rate_threshold=3):
    timestamps = [log["time"][:5] for log in logs] # group by minute
    counts = Counter(timestamps)

    for minute, count in counts.items():
        if count >= rate_threshold:
            print(f"[CRITICAL ALERT] Rapid file modification at {minute}: {count} files changed!")

    for log in logs:
        if log["ext"] in ransom_exts:
            print(f"[RANSOMWARE EXTENSION DETECTED] File renamed to: {log['file']}")

detect_ransomware_behavior(file_logs, rate_threshold=3)
```

**Result**

Successfully flagged ransomware-like behavior based on rapid file renaming rates and suspicious extension changes.
