**Aim**

To analyze domain query logs and identify suspicious domain names using basic domain reputation indicators and typosquatting detection.

**Algorithm**

Define a list of legitimate reference domains (e.g., paypal.com, google.com).

Load queried domain names from network DNS logs.

Check for high-risk top-level domains (e.g., .top, .xyz, .cc).

Identify suspicious domain keywords (e.g., login, verify, update, secure).

Calculate similarity or keyword overlap to detect typosquatting.

Flag domain names exhibiting excessive length or excessive subdomains.

Group flagged suspicious domain query entries.

Output domain reputation analysis report.

```python
queried_domains = [
    "google.com",
    "paypa1-security-update.xyz",
    "secure-login-bank.top",
    "github.com"
]

suspicious_tlds = [".xyz", ".top"]
suspicious_keywords = ["paypa1", "security-update", "secure-login"]

def analyze_domains(domains):
    print("--- Suspicious Domain Reputation Analysis ---")
    for domain in domains:
        if any(domain.endswith(tld) for tld in suspicious_tlds) or any(kw in domain for kw in suspicious_keywords):
            print(f"[SUSPICIOUS DOMAIN] {domain}")

analyze_domains(queried_domains)
```

**Result**

Identified malicious and typosquatted domains using domain reputation rules and suspicious keyword analysis.
