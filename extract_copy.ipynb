**Aim:**

To create an exact, byte-for-byte clone of a digital file and verify that the copy is perfectly identical to the original using cryptographic hashing (SHA-256), maintaining a strict chain of custody.

**Algorithm :**

Start the program.

Import the hashlib library.

Define a function to calculate the SHA-256 hash of a file by reading it in 4KB chunks.

Create a sample "original" file with dummy text for testing purposes.

Open the original file in binary read mode (rb) and the destination file in binary write mode (wb).

Read chunks from the source file and write them sequentially to the destination file.

Compute the SHA-256 hashes for both the original file and the newly copied file.

Compare the two hashes; if they match exactly, output success, otherwise report a failure.

```python
import hashlib
import os

def calculate_sha256(file_path):
    sha256_hash = hashlib.sha256()
    with open(file_path, "rb") as f:
        for byte_block in iter(lambda: f.read(4096), b""):
            sha256_hash.update(byte_block)
    return sha256_hash.hexdigest()

def bitstream_copy(src, dest):
    print(f"Starting bit-stream copy from {src} to {dest}...")
    with open(src, "rb") as source_file, open(dest, "wb") as dest_file:
        for byte_block in iter(lambda: source_file.read(4096), b""):
            dest_file.write(byte_block)

    hash_original = calculate_sha256(src)
    hash_copy = calculate_sha256(dest)

    print(f"Original SHA-256: {hash_original}")
    print(f"Copy SHA-256:     {hash_copy}")

    if hash_original == hash_copy:
        print("Integrity Verified: The hashes match. Forensic copy is exact.")
    else:
        print("Integrity Failed: The hashes do not match.")

# Create a sample file
original_file = "evidence_original.txt"
copied_file = "evidence_clone.dd"
with open(original_file, "w") as f:
    f.write("CONFIDENTIAL DATA: Top secret project details inside.")

bitstream_copy(original_file, copied_file)
```

**Result:**

A duplicated file is generated alongside two calculated SHA-256 hash values. The identical hash outputs confirm that zero bits were altered during the copying process, ensuring the evidence remains forensically sound.
