**Aim**

To write a Python program that programmatically generates and formats a structured text report detailing the five key stages of the digital forensic investigation lifecycle (Identification, Preservation, Collection, Analysis, and Reporting) for documentation and audit purposes.

**Algorithm**

Start

Define Data Structure: Create a dictionary or list of objects containing information for each of the five lifecycle stages (Stage Name, Description, and Key Activities).

Initialize Report Builder: Open a target file (forensic_report.txt) in write mode ("w") or build a string in memory using standard formatting.

Generate Header: Append a report title, timestamp/date, and metadata header.

Iterate Through Stages: Loop through each stage in the data structure:

Format stage titles using clear structural headings.

Print/write the stage description and bulleted key activities.

Generate Footer: Append an overall summary or sign-off section at the end of the report.

Save & Output: Write the structured report to disk and print a confirmation message to the console.

End

```python
import datetime

def generate_forensic_report(output_filename="forensic_lifecycle_report.txt"):

    lifecycle_stages = [
        {
            "stage": "1. Identification",
            "description": "Recognizing and isolating potential digital evidence from devices and cloud environments.",
            "activities": [
                "Identify incident scope and affected assets",
                "Locate physical and virtual storage media",
                "Determine required search warrants or legal authorizations"
            ]
        },
        {
            "stage": "2. Preservation",
            "description": "Securing digital evidence to prevent alteration, contamination, or destruction.",
            "activities": [
                "Isolate devices from networks (Faraday bags / flight mode)",
                "Document chain of custody records",
                "Perform write-blocking before connecting media"
            ]
        },
        {
            "stage": "3. Collection",
            "description": "Acquiring forensic copies of evidence in a standardized, legally defensible manner.",
            "activities": [
                "Create bit-stream physical disk images (e.g., RAW/E01)",
                "Calculate and record cryptographic hashes (SHA-256)",
                "Capture volatile RAM data prior to system shutdown"
            ]
        },
        {
            "stage": "4. Analysis",
            "description": "Examining collected digital artifacts to reconstruct events and extract evidence.",
            "activities": [
                "Parse system logs, registry files, and file system metadata",
                "Recover deleted files and unallocated space",
                "Correlate timeline events and user activities"
            ]
        },
        {
            "stage": "5. Reporting",
            "description": "Documenting findings clearly for presentation to legal authority or management.",
            "activities": [
                "Compile detailed, objective investigative reports",
                "Include original hash values and chain of custody logs",
                "Prepare expert witness testimony based on verified facts"
            ]
        }
    ]


    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    report_lines = []
    report_lines.append("=========================================================")
    report_lines.append("     DIGITAL FORENSIC INVESTIGATION LIFECYCLE REPORT     ")
    report_lines.append("=========================================================")
    report_lines.append(f"Generated On : {timestamp}")
    report_lines.append("Status       : Standard Operating Procedure (SOP)")
    report_lines.append("---------------------------------------------------------\n")

    for item in lifecycle_stages:
        report_lines.append(f"STAGE: {item['stage']}")
        report_lines.append(f"Summary: {item['description']}")
        report_lines.append("Key Activities:")
        for activity in item["activities"]:
            report_lines.append(f"  - {activity}")
        report_lines.append("\n" + "-" * 57 + "\n")

    report_lines.append("=================== END OF REPORT ===================")


    full_report = "\n".join(report_lines)


    with open(output_filename, "w") as f:
        f.write(full_report)


    print(full_report)
    print(f"\n Report successfully generated and saved to: {output_filename}")

if __name__ == "__main__":
    generate_forensic_report()
```

**Result:**

The program successfully compiles the provided case inputs into a highly structured text file. This standardized document guarantees that all critical phases of the forensic lifecycle are clearly articulated for court or stakeholder review.
