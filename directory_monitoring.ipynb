**Aim**

To monitor a designated directory and record real-time file creation, modification, deletion, and access events.

**Algorithm**

Import the os and time modules.

Define the path of the directory to be monitored.

Fetch the initial list of files and their last modification timestamps.

Store the initial directory state in a dictionary baseline.

Enter a monitoring loop with a periodic sleep delay.

Check for newly added files in the directory state.

Identify missing files relative to the baseline as deleted.

Detect modified files by comparing current modification timestamps with baseline records.

```python
import os
import time

def monitor_directory(path, duration_sec=5):
    print(f"[*] Monitoring directory: {path}")
    baseline = {f: os.path.getmtime(os.path.join(path, f)) for f in os.listdir(path)}

    end_time = time.time() + duration_sec
    while time.time() < end_time:
        time.sleep(1)
        current_files = {f: os.path.getmtime(os.path.join(path, f)) for f in os.listdir(path)}

        for f in current_files:
            if f not in baseline:
                print(f"[EVENT] Created: {f}")
            elif current_files[f] != baseline[f]:
                print(f"[EVENT] Modified: {f}")

        for f in baseline:
            if f not in current_files:
                print(f"[EVENT] Deleted: {f}")

        baseline = current_files

# Execute on current folder
monitor_directory(".", duration_sec=2)
```

**Result**

Monitored the specified directory and successfully detected file creation, modification, and deletion activity events.
