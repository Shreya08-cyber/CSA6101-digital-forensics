**Aim**

To analyze network flow records and identify anomalous outbound data transfers indicative of data exfiltration.

**Algorithm**

Load simulated outbound network traffic flow records (Internal IP, External IP, Data Transferred Bytes).

Aggregate total bytes transmitted per internal host IP.

Calculate baseline or mean outbound transfer volume.

Set anomaly threshold (e.g., transfer > 50 MB).

Identify internal hosts transmitting data exceeding the threshold.

Inspect destination ports used for high-volume transfers.

Flag abnormal outbound data spikes.

Output data exfiltration alert report.

```python
  traffic_records = [
    {"src": "192.168.1.100", "dst": "93.184.216.34", "bytes": 1500},
    {"src": "192.168.1.105", "dst": "198.51.100.22", "bytes": 104857600}, # ~100MB
    {"src": "192.168.1.100", "dst": "172.217.16.206", "bytes": 2048}
]

def detect_exfiltration(records, threshold_bytes=10000000): # 10MB
    print("--- Outbound Network Anomaly Report ---")
    for rec in records:
        if rec["bytes"] > threshold_bytes:
            mb = rec["bytes"] / (1024 * 1024)
            print(f"[EXFILTRATION ALERT] Host {rec['src']} uploaded {mb:.2f} MB to External IP: {rec['dst']}")

detect_exfiltration(traffic_records)
```

**Result**

Detected unusual outbound data spikes, highlighting potential data exfiltration from internal systems.
