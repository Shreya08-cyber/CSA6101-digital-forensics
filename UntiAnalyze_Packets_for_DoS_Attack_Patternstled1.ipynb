**Aim**

To process packet traffic metrics and detect Denial-of-Service (DoS) attack flooding patterns.

**Algorithm**

Load packet capture metrics (Timestamp, Source IP, Protocol, Packet Count).

Group packet traffic by Source IP address within time intervals.

Calculate packet rate per second/minute for each source IP.

Set threshold for high-volume flooding alert (e.g., > 500 packets/sec).

Detect ICMP, SYN, or UDP traffic volume anomalies.

Identify top talkers sending massive packet floods.

Flag offending Source IPs attempting DoS attacks.

Display DoS attack pattern summary report.

```python
from collections import defaultdict

packet_logs = [
    {"src": "172.16.0.5", "proto": "TCP-SYN", "count": 200},
    {"src": "172.16.0.5", "proto": "TCP-SYN", "count": 400},
    {"src": "10.0.0.2", "proto": "HTTP", "count": 5}
]

def detect_dos_patterns(logs, threshold=500):
    traffic = defaultdict(int)
    for pkt in logs:
        traffic[pkt["src"]] += pkt["count"]

    print("--- DoS Attack Pattern Detection ---")
    for ip, total_packets in traffic.items():
        if total_packets >= threshold:
            print(f"[DoS ALERT] High volume packet flood from IP: {ip} ({total_packets} packets)")

detect_dos_patterns(packet_logs, threshold=500)
```

**Result**

Detected DoS attack patterns by identifying high-volume packet flooding sources from network traffic logs.
