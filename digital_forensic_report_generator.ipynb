**Aim**

To automatically compile digital investigation findings into a structured, standardized digital forensic report formatted in Markdown or plain text

**Algorithm**

Gather core investigation metadata (case ID, investigator name, date, evidence hash).

Gather technical findings (e.g., suspicious files, IP addresses, log evidence).

Define a structured report template containing standard forensic sections (Header, Evidence, Findings, Conclusion).

Create a function to format key-value pairs into structured Markdown string blocks.

Inject case metadata dynamically into the report template.

Format findings into clean bullet points or tables.

Append investigator sign-off and integrity checksum placeholders.

Save the populated forensic report into a file on disk (e.g., forensic_report.md).

```python
import datetime
import os

def generate_forensic_report(case_id, investigator, evidence_name, findings, file_sha256):
    report_md = f"""# DIGITAL FORENSIC INVESTIGATION REPORT

## 1. Executive Summary
- **Case ID:** {case_id}
- **Lead Investigator:** {investigator}
- **Report Generated:** {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S IST')}
- **Status:** Closed / Findings Attached

---

## 2. Evidence Registry
| Field | Value |
|---|---|
| **Evidence ID** | EV-001 |
| **Source Name** | {evidence_name} |
| **Integrity Hash (SHA-256)** | `{file_sha256}` |

---

## 3. Key Technical Findings
"""
    for idx, item in enumerate(findings, 1):
        report_md += f"{idx}. {item}\n"

    report_md += """
---

## 4. Conclusion & Chain of Custody
Analysis confirms unauthorized administrative access via SSH brute-force vector.
Chain of custody maintained verified under standard forensic principles.
"""

    output_filename = f"Forensic_Report_{case_id}.md"

    with open(output_filename, "w") as report_file:
        report_file.write(report_md)

    print(f"[SUCCESS] Forensic report generated: {output_filename}\n")


    print("--- GENERATED FILE CONTENT PREVIEW ---")
    print(report_md)

findings_list = [
    "Unusual SSH access from unauthorized external IP 192.168.1.50 at 03:14:02 AM.",
    "Multiple failed login attempts recorded within a 20-second window preceding entry.",
    "Staging directory created at `/tmp/.exfil` containing compressed user archives."
]

generate_forensic_report(
    case_id="DF-2026-009",
    investigator="Det. Vance",
    evidence_name="Server_Disk_Image.img",
    findings=findings_list,
    file_sha256="e3b0c44298fc1c149afbf4c8996fb92427ae41e4649b934ca495991b7852b855"
)
```

**Result**

The generator creates a complete Markdown report document on the Colab filesystem (Forensic_Report_DF-2026-009.md). It populates real-time timestamps, evidence hashes, and ordered findings into a clean format ready for legal or technical review.
