**Aim**

To scan simulated Windows USB connection logs and detect unauthorized USB device insertions.

**Algorithm**

Define a baseline list of authorized USB device hardware IDs.

Supply simulated USB log records containing Device ID, Serial, Timestamp, and Action.

Filter log entries representing USB insertion/connection events.

Extract Device ID and serial details from connection events.

Compare extracted Device ID against authorized hardware baseline.

Flag any connection from non-whitelisted USB hardware.

Record event timestamp and device details for compliance auditing.

Output summary of unauthorized USB connection alerts.

```python
authorized_usb_ids = ["USB\\VID_0781&PID_5581", "USB\\VID_0951&PID_1666"]

usb_logs = [
    {"timestamp": "2026-08-06 09:00", "device_id": "USB\\VID_0781&PID_5581", "action": "Connected"},
    {"timestamp": "2026-08-06 09:45", "device_id": "USB\\VID_9999&PID_8888", "action": "Connected"}
]

def audit_usb_connections(logs, whitelist):
    print("--- USB Connection Audit ---")
    for log in logs:
        if log["action"] == "Connected":
            if log["device_id"] not in whitelist:
                print(f"[ALERT {log['timestamp']}] Unauthorized USB Device: {log['device_id']}")
            else:
                print(f"[OK {log['timestamp']}] Authorized USB Device: {log['device_id']}")

audit_usb_connections(usb_logs, authorized_usb_ids)
```

**Result**

Detected unauthorized USB hardware connections by comparing hardware IDs against an approved whitelist.
