**Aim**

To aggregate file timestamps and system log records into a unified, chronologically ordered digital event timeline.

**Algorithm**

Import datetime module.

Collect file timestamp records (Created, Modified, Accessed).

Collect system event logs with timestamps.

Normalize all timestamp strings into standard datetime objects.

Combine events into a master timeline list.

Sort the combined list chronologically by timestamp.

Format output to show timestamp, event source, and event description.

Output the generated forensic event timeline.

```python
from datetime import datetime

file_events = [
    {"time": "2026-08-06 10:15:00", "source": "FileSystem", "desc": "file_deleted: evidence.txt"},
    {"time": "2026-08-06 10:00:00", "source": "FileSystem", "desc": "file_created: draft.docx"}
]

sys_logs = [
    {"time": "2026-08-06 10:05:00", "source": "SysLog", "desc": "User admin logged in"}
]

def generate_timeline(file_logs, system_logs):
    timeline = file_logs + system_logs
    timeline.sort(key=lambda x: datetime.strptime(x["time"], "%Y-%m-%d %H:%M:%S"))

    print("--- Unified Forensic Timeline ---")
    for event in timeline:
        print(f"[{event['time']}] [{event['source']}] {event['desc']}")

generate_timeline(file_events, sys_logs)
```

**Result**

Generated a chronologically sorted forensic timeline merging file system events and system activity logs.
