**Aim**
  
To parse simulated ransomware incident logs and reconstruct the step-by-step attack sequence chronologically.

**Algorithm**

Load raw ransomware event log records.

Standardize event fields (timestamp, stage, action, process ID).

Sort event entries chronologically.

Identify initial execution phase (e.g., macro execution/phishing payload).

Trace persistence creation phase (e.g., registry run key added).

Map bulk encryption trajectory across directories.

Identify ransom note creation event (e.g., READ_ME.txt).

Print structured attack sequence report.

```python
incident_logs = [
    {"time": "09:00:00", "stage": "Initial Access", "desc": "Malicious attachment executed from Outlook"},
    {"time": "09:00:05", "stage": "Persistence", "desc": "Registry key added: HKCU\\Software\\Microsoft\\Run"},
    {"time": "09:01:00", "stage": "Execution", "desc": "VSSAdmin shadow copies deleted"},
    {"time": "09:02:00", "stage": "Encryption", "desc": "Bulk encryption started in C:\\Documents"},
    {"time": "09:03:00", "stage": "Impact", "desc": "Ransom note created: HOW_TO_DECRYPT.txt"}
]

def reconstruct_attack(logs):
    logs.sort(key=lambda x: x["time"])
    print("=== Ransomware Attack Sequence Reconstruction ===")
    for idx, log in enumerate(logs, 1):
        print(f"Step {idx} [{log['time']}] Phase: {log['stage']} -> {log['desc']}")

reconstruct_attack(incident_logs)
```

**Result**

Reconstructed the chronological ransomware attack sequence from initial entry to encryption and ransom note creation.
