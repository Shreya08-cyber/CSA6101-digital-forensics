**Aim**

To process firewall log records and identify blocked connections and repeated unauthorized access attempts.

**Algorithm**

Load firewall log data (Timestamp, Source IP, Destination Port, Action).

Filter entries where Action is equal to BLOCKED or DENIED.

Extract offending Source IPs from filtered entries.

Count total blocked connection attempts per Source IP.

Identify targeted ports (e.g., SSH 22, RDP 3389) associated with blocked attempts.

Set threshold for port scanning or brute-force alerts.

Flag Source IPs exceeding connection block threshold.

Display firewall breach risk summary.

```python
from collections import Counter

firewall_logs = [
    {"src": "10.0.0.50", "port": 22, "action": "BLOCKED"},
    {"src": "10.0.0.50", "port": 22, "action": "BLOCKED"},
    {"src": "10.0.0.50", "port": 22, "action": "BLOCKED"},
    {"src": "192.168.1.5", "port": 80, "action": "ALLOWED"}
]

def analyze_firewall(logs, block_threshold=2):
    blocked_ips = [log["src"] for log in logs if log["action"] == "BLOCKED"]
    counts = Counter(blocked_ips)

    print("--- Firewall Blocked Connections Summary ---")
    for ip, count in counts.items():
        if count >= block_threshold:
            print(f"[ALERT] Repeated unauthorized attempts from IP: {ip} ({count} blocked attempts)")

analyze_firewall(firewall_logs, block_threshold=2)
```

**Result**

Identified repeated unauthorized access attempts and blocked connection hotspots from firewall logs.
