**Aim**

To compare IP addresses from network security logs against threat intelligence indicator feeds to identify malicious connections.

**Algorithm**

Define a list of known malicious IP threat indicators.

Load network connection log records (Timestamp, Source IP, Destination IP, Port).

Extract IP addresses from log records.

Cross-reference connection IPs against threat intelligence lists.

Count total interaction attempts with malicious IPs.

Identify affected internal source hosts contacting flagged IPs.

Categorize threat level based on connection frequency.

Output threat detection report containing matched IP addresses.

```python
threat_intel_ips = {"198.51.100.14", "203.0.113.55"}

network_logs = [
    {"src": "192.168.1.10", "dst": "8.8.8.8", "port": 53},
    {"src": "192.168.1.15", "dst": "198.51.100.14", "port": 443},
    {"src": "192.168.1.20", "dst": "203.0.113.55", "port": 8080}
]

def analyze_network_ips(logs, threat_feed):
    print("--- Threat Intelligence Matching ---")
    for log in logs:
        if log["dst"] in threat_feed:
            print(f"[ALERT] Internal host {log['src']} connected to Malicious IP: {log['dst']} on Port {log['port']}")

analyze_network_ips(network_logs, threat_intel_ips)
```

**Result**

Flagged malicious IP address connections by cross-referencing network logs with threat indicator feeds.
