**Aim**

To parse simulated system activity logs and identify files accessed within a recent specific timeframe.

**Algorithm**

Import datetime and timedelta modules.

Define simulated system activity logs with timestamps and accessed file paths.

Parse string timestamps into native Python datetime objects.

Define a time window threshold (e.g., access within the last 2 hours).

Fetch current or simulated reference baseline time.

Iterate through log entries and compare event timestamps against threshold.

Filter entries that fall within the specified recent timeframe.

Display chronologically sorted recently accessed file entries.

```python
from datetime import datetime, timedelta

activity_logs = [
    {"file": "/docs/plan.pdf", "timestamp": "2026-08-06 10:30:00"},
    {"file": "/etc/passwd", "timestamp": "2026-08-06 09:15:00"},
    {"file": "/home/user/notes.txt", "timestamp": "2026-08-06 07:00:00"}
]

def extract_recent_files(logs, hours=2):
    ref_time = datetime.strptime("2026-08-06 11:00:00", "%Y-%m-%d %H:%M:%S")
    threshold = ref_time - timedelta(hours=hours)

    print(f"--- Files Accessed After {threshold} ---")
    for entry in logs:
        log_time = datetime.strptime(entry["timestamp"], "%Y-%m-%d %H:%M:%S")
        if log_time >= threshold:
            print(f"[{entry['timestamp']}] {entry['file']}")

extract_recent_files(activity_logs, hours=2)
```

**Result**

Extracted recently accessed files from system log data based on defined time window parameters.
