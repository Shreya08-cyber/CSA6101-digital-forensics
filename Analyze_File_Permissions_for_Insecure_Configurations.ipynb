**Aim**

To inspect file permissions in a directory and flag files configured with insecure permissions (e.g., world-writable or overly permissive access).

**Algorithm**

Import os and stat modules.

Define directory path to analyze.

List files in specified directory.

Fetch permissions bitmask using os.stat().st_mode.

Extract permissions string (octal representation, e.g., 777).

Check if file is world-writable (stat.S_IWOTH).

Check if file is world-readable (stat.S_IROTH).

Flag files violating security permission baselines.

```python
import os
import stat

def analyze_file_permissions(folder_path):
    print(f"--- Permissive File Analysis: {folder_path} ---")
    for root, _, files in os.walk(folder_path):
        for file in files:
            path = os.path.join(root, file)
            try:
                mode = os.stat(path).st_mode
                octal_perm = oct(mode)[-3:]

                if mode & stat.S_IWOTH:
                    print(f"[INSECURE] World-Writable ({octal_perm}): {path}")
                elif mode & stat.S_IROTH:
                    print(f"[WARNING] World-Readable ({octal_perm}): {path}")
            except Exception:
                continue

analyze_file_permissions(".")
```

**Result**

Analyzed file system permissions and identified files configured with insecure world-writable access modes.
