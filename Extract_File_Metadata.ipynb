**Aim**
  
To extract core forensic file metadata, including file name, size, creation time, modification time, and last access time.

**Algorithm**

Import the os and datetime modules.

Specify the target file path for metadata extraction.

Check if the specified file exists on disk.

Retrieve file statistics using os.stat().

Read file size in bytes and convert to kilobytes.

Extract creation time (st_ctime), modification time (st_mtime), and access time (st_atime).

Convert raw timestamps into human-readable date and time strings.

Output formatted file metadata details.

```python
import os
from datetime import datetime

def extract_metadata(file_path):
    if not os.path.exists(file_path):
        print("File not found!")
        return

    stats = os.stat(file_path)
    print(f"File Name:        {os.path.basename(file_path)}")
    print(f"File Size:        {stats.st_size} bytes")
    print(f"Creation Time:    {datetime.fromtimestamp(stats.st_ctime)}")
    print(f"Modification Time:{datetime.fromtimestamp(stats.st_mtime)}")
    print(f"Access Time:      {datetime.fromtimestamp(stats.st_atime)}")

extract_metadata("evidence.txt")
```

**Result**

Extracted complete file metadata including filename, size, and MAC (Modified, Accessed, Created) timestamps.
