**Aim**

To calculate and compare MD5, SHA-1, and SHA-256 cryptographic hash values for a digital evidence file to verify data integrity and detect tampering.

**Algorithm**

Import the built-in hashlib module in Python.

Specify the path of the target digital evidence file.

Initialize MD5, SHA-1, and SHA-256 hashing objects.

Open the evidence file in binary read mode (rb).

Read the file content in fixed-size binary chunks (e.g., 4096 bytes).

Update each hash object iteratively with the read data chunks.

Compute the hexadecimal digest for MD5, SHA-1, and SHA-256.

Display the generated hash values for cryptographic comparison.

```python
import hashlib

def calculate_file_hashes(file_path):
    md5_hash = hashlib.md5()
    sha1_hash = hashlib.sha1()
    sha256_hash = hashlib.sha256()

    with open(file_path, "rb") as f:
        for chunk in iter(lambda: f.read(4096), b""):
            md5_hash.update(chunk)
            sha1_hash.update(chunk)
            sha256_hash.update(chunk)

    print(f"File: {file_path}")
    print(f"MD5:    {md5_hash.hexdigest()}")
    print(f"SHA-1:  {sha1_hash.hexdigest()}")
    print(f"SHA-256:{sha256_hash.hexdigest()}")

# Create sample file for testing
sample_file = "evidence.txt"
with open(sample_file, "w") as f:
    f.write("Digital Evidence Content - Case #1024")

calculate_file_hashes(sample_file)
```

**Result**

Successfully calculated MD5, SHA-1, and SHA-256 hash values, establishing an immutable cryptographic baseline for the evidence file.
