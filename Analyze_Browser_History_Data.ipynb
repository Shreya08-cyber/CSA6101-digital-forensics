**Aim**
To process browser history logs, calculate domain visit frequencies, and identify potentially suspicious URLs.

**Algorithm**

Import re and Counter from collections.

Supply simulated browser history log records containing visited URLs.

Parse host domain names from full URLs using regular expressions.

Count occurrence frequency for each extracted domain name.

Define security rules and suspicious indicators (e.g., TLDs like .xyz, keywords like phish, free-gift).

Iterate through browser history records.

Flag URLs matching defined suspicious threat keywords or high-risk domains.

Output top visited websites along with flagged suspicious URLs.

```python
import re
from collections import Counter

history_logs = [
    "https://google.com/search?q=forensics",
    "https://github.com/login",
    "http://free-gift-card.xyz/claim",
    "https://google.com/mail",
    "http://login-verify-bank.com/secure"
]

def analyze_browser_history(urls):
    domains = [re.sub(r'https?://([^/]+).*', r'\1', url) for url in urls]
    freq = Counter(domains)

    print("--- Top Visited Domains ---")
    for domain, count in freq.most_common():
        print(f"{domain}: {count} visit(s)")

    print("\n--- Flagged Suspicious URLs ---")
    suspicious_keywords = ["free-gift", "login-verify", ".xyz"]
    for url in urls:
        if any(kw in url for kw in suspicious_keywords):
            print(f"[SUSPICIOUS] {url}")

analyze_browser_history(history_logs)
```

**Result**

Analyzed browser history logs to generate domain frequency rankings and detect malicious phishing URLs.
