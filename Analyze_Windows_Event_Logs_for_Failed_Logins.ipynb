**Aim**

To scan simulated Windows Event Logs (Event ID 4625) and detect brute-force password attacks by identifying repeated failed login attempts.

**Algorithm**

Import Counter from collections.

Define simulated event log records containing Event ID, User, and Source IP.

Filter log entries matching Event ID 4625 (Failed Logon).

Extract user accounts and source IP addresses from matching logs.

Aggregate failed logon attempt counts per user and IP combination.

Set threshold for brute-force alert trigger (e.g., > 2 failures).

Flag accounts and IP addresses exceeding failure threshold.

Output detailed report on detected brute-force login attempts.

```python
from collections import Counter

event_logs = [
    {"event_id": 4625, "user": "admin", "ip": "192.168.1.50"},
    {"event_id": 4625, "user": "admin", "ip": "192.168.1.50"},
    {"event_id": 4625, "user": "admin", "ip": "192.168.1.50"},
    {"event_id": 4624, "user": "john", "ip": "192.168.1.12"},
    {"event_id": 4625, "user": "root", "ip": "10.0.0.5"}
]

def detect_failed_logins(logs, threshold=2):
    failed_attempts = [f"{log['user']} from {log['ip']}" for log in logs if log['event_id'] == 4625]
    counts = Counter(failed_attempts)

    print("--- Repeated Failed Login Alerts ---")
    for target, count in counts.items():
        if count >= threshold:
            print(f"[ALERT] Brute-force suspicious: {target} - Failed {count} times!")

detect_failed_logins(event_logs, threshold=2)
```

**Result**

Detected potential brute-force attacks by identifying repeated Windows Event ID 4625 failed login entries.
