**Aim**

To extract embedded Exchangeable Image File Format (EXIF) metadata (such as camera model, date taken, and software used) from a JPEG/TIFF image file using Python.

**Algorithm**

Import the necessary image processing library (PIL.Image and PIL.ExifTags).

Specify the filepath of the target image file.

Open the image file using Image.open().

Call the _getexif() method to extract raw metadata tags.

Check if valid EXIF metadata exists within the file.

Map numerical EXIF tag keys to human-readable strings using ExifTags.TAGS.

Iterate through the key-value metadata pairs and format them.

Output the extracted metadata key-value details to the terminal.

```python
from PIL import Image, ExifTags

img_path = "sample_exif_image.jpg"
image = Image.new("RGB", (200, 200), color="blue")

exif_obj = image.getexif()
exif_obj[271] = "Sample Camera Corp"  # Tag 271 = Make
exif_obj[272] = "Model X-2026"         # Tag 272 = Model
exif_obj[305] = "Colab Forensic Engine"# Tag 305 = Software

image.save(img_path, exif=exif_obj)
print(f"Created sample image at: {img_path}\n")

def extract_exif_metadata(file_path):
    try:
        img = Image.open(file_path)
        exif_data = img._getexif()

        if not exif_data:
            print("[INFO] No EXIF metadata found in the image.")
            return

        print(f"--- EXIF Metadata for {file_path} ---")
        for tag_id, value in exif_data.items():
            tag_name = ExifTags.TAGS.get(tag_id, f"Unknown (ID: {tag_id})")
            print(f"{tag_name:<20}: {value}")

    except Exception as e:
        print(f"[ERROR] Could not read image EXIF data: {e}")

extract_exif_metadata(img_path)
```

**Result**

The code constructs a sample JPEG file containing custom EXIF tags directly in Colab's local filesystem. When executed, the extraction function parses the binary metadata and outputs clear tag mappings (Make, Model, Software) to the screen.
