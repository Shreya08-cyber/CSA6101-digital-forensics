**Aim**

To identify duplicate digital evidence files across directories by computing and comparing their SHA-256 hash values.

**Algorithm**

Import the os and hashlib modules.

Define target directory path to scan for duplicate files.

Initialize a dictionary to store SHA-256 hashes as keys and file path lists as values.

Walk through all files in the target folder.

Compute the SHA-256 hash for every encountered file.

Append file paths to matching hash entries in the dictionary.

Filter entries that contain more than one associated file path.

Output grouped duplicate files sharing identical SHA-256 signatures.

```python
import os
import hashlib
from collections import defaultdict

def find_duplicate_files(folder_path):
    hashes = defaultdict(list)
    for root, _, files in os.walk(folder_path):
        for file in files:
            file_path = os.path.join(root, file)
            try:
                with open(file_path, "rb") as f:
                    file_hash = hashlib.sha256(f.read()).hexdigest()
                    hashes[file_hash].append(file_path)
            except Exception:
                continue

    duplicates = {h: paths for h, paths in hashes.items() if len(paths) > 1}
    for h, paths in duplicates.items():
        print(f"[DUPLICATE FOUND] SHA-256: {h[:16]}...")
        for p in paths:
            print(f"  -> {p}")

find_duplicate_files(".")
```

**Result**

Identified duplicate digital evidence files by grouping identical SHA-256 hash signatures across the target directory.
