**Aim**
To analyze system login attempt logs and flag potential brute-force intrusion attempts by tracking failed authentication threshold breaches originating from the same source IP address.

**Algorithm**

Open and read the target system login log file.

Define a threshold for maximum allowable failed attempts (e.g., 3 failed attempts).

Initialize a data structure (such as a dictionary) to keep track of failed attempts per IP address.

Iterate line by line through the log entries.

Parse out the IP address and the login status (FAILED vs SUCCESS).

Increment the counter for the IP address if the status indicates a failure.

Evaluate if the failure count for any IP exceeds the defined threshold.

Output an alert message indicating a suspected brute-force attack along with the offending IP.

```python
# 1. Create a sample log file in Colab
log_filename = "system_log.txt"

sample_logs = [
    "2026-07-30 10:00:01 IP: 192.168.1.50 Status: FAILED\n",
    "2026-07-30 10:00:05 IP: 192.168.1.50 Status: FAILED\n",
    "2026-07-30 10:00:10 IP: 192.168.1.50 Status: FAILED\n",
    "2026-07-30 10:00:12 IP: 10.0.0.12 Status: SUCCESS\n",
    "2026-07-30 10:00:15 IP: 192.168.1.50 Status: FAILED\n",
    "2026-07-30 10:00:20 IP: 10.0.0.12 Status: FAILED\n"
]

with open(log_filename, "w") as f:
    f.writelines(sample_logs)

# 2. Process and Analyze Log File
THRESHOLD = 3
failed_attempts = {}

print("--- LOG ANALYSIS REPORT ---")

with open(log_filename, "r") as f:
    for line in f:
        parts = line.strip().split()
        if len(parts) >= 5:
            ip = parts[2]
            status = parts[4]
            print(f"Processing Entry -> IP: {ip:<12} Status: {status}")

            if status == "FAILED":
                failed_attempts[ip] = failed_attempts.get(ip, 0) + 1

print("-" * 35)
print("--- SUMMARY ALERTS ---")

# 3. Check for Threshold Breaches
alert_triggered = False
for ip, count in failed_attempts.items():
    if count >= THRESHOLD:
        print(f"[ALERT] Potential Brute-Force attack detected from IP: {ip} ({count} failed attempts)")
        alert_triggered = True
    else:
        print(f"[INFO] IP: {ip} has {count} failed attempt(s). Within safe limits.")

if not alert_triggered:
    print("[INFO] No brute-force attacks detected.")
```
