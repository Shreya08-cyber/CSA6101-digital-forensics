**Aim**

To detect unauthorized file modifications by comparing current file hashes against a pre-established baseline hash catalog.

**Algorithm**

Import hashlib and os modules.

Define baseline hashes for critical evidence files in a dictionary.

Scan the target directory for existing files.

Calculate the current SHA-256 hash for each file.

Compare the calculated SHA-256 hash against the baseline entry.

Flag files whose computed hashes do not match baseline records.

Identify missing files present in the baseline but absent on disk.

Display an alert report highlighting unauthorized file alterations.

```python
import hashlib
import os

baseline_hashes = {
    "evidence.txt": "a823e51f4bfbc50bd82b947c61f2f8115682cb380e227916960d7ca0bc1620ef"
}

def verify_file_integrity(baseline):
    for filename, expected_hash in baseline.items():
        if not os.path.exists(filename):
            print(f"[ALERT] File missing: {filename}")
            continue

        with open(filename, "rb") as f:
            current_hash = hashlib.sha256(f.read()).hexdigest()

        if current_hash == expected_hash:
            print(f"[OK] {filename}: Integrity verified.")
        else:
            print(f"[CRITICAL] {filename}: Unauthorized modification detected!")

verify_file_integrity(baseline_hashes)
```

**Result**

Detected unauthorized file modifications by identifying checksum mismatches against stored baseline hash values.
