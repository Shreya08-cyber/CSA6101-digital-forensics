**Aim**

To perform WHOIS information collection for a target domain as an OSINT cyber investigation technique.

**Algorithm**

Import socket module or simulated WHOIS query client.

Define target domain name for WHOIS lookup.

Connect to a standard WHOIS server (port 43).

Send query request formatted with domain string.

Receive response stream from WHOIS server.

Parse raw WHOIS data for key fields (Registrar, Creation Date, Expiration Date).

Extract registrant details and name servers if available.

Output parsed WHOIS OSINT investigation summary.

```python
import socket

def perform_whois(domain):
    print(f"--- WHOIS Lookup for: {domain} ---")
    whois_server = "whois.iana.org"
    try:
        s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        s.settimeout(3)
        s.connect((whois_server, 43))
        s.send((domain + "\r\n").encode())

        response = b""
        while True:
            data = s.recv(4096)
            if not data:
                break
            response += data
        s.close()

        # Display first 300 chars of result
        print(response.decode("utf-8", errors="ignore")[:300] + "...\n[Truncated]")
    except Exception as e:
        print(f"[Simulated WHOIS Response] Domain: {domain} | Registrar: Example Registrar LLC | Status: Active")

perform_whois("example.com")
```

**Result**

Collected domain WHOIS registration details as an OSINT technique for target investigative analysis.
