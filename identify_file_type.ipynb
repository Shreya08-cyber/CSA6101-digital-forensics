**Aim:**

To determine the actual format of a file by reading its raw hexadecimal header (magic numbers), ignoring its file extension. This helps investigators defeat anti-forensics techniques where suspects rename extensions to hide data.

**Algorithm:**

Start the program.

Define a dictionary mapping known hexadecimal signatures to file types (e.g., 25504446 for PDF).

Create a test file disguised with a fake extension for demonstration.

Open the target file in binary read mode (rb).

Read the first 8 bytes (the file header).

Convert the extracted bytes into a hexadecimal string representation and convert it to uppercase.

Iterate through the dictionary to see if the hex string starts with any known magic numbers.

Print the matched file type or "Unknown format" if no match is found.

```python
def check_magic_numbers(file_path):

    signatures = {
        "25504446": "PDF Document",
        "FFD8FFE0": "JPEG Image",
        "89504E47": "PNG Image",
        "504B0304": "ZIP Archive"
    }

    with open(file_path, "rb") as f:

        header = f.read(4)

    hex_header = header.hex().upper()
    print(f"File Header (Hex): {hex_header}")

    for sig, file_type in signatures.items():
        if hex_header.startswith(sig):
            return file_type

    return "Unknown/Unrecognized format"


disguised_file = "secret_image.txt"
with open(disguised_file, "wb") as f:
    f.write(b'\x25\x50\x44\x46\x2D\x31\x2E\x33')

real_type = check_magic_numbers(disguised_file)
print(f"Analyzed File: {disguised_file}")
print(f"True File Type: {real_type}")
```

**Result:**

The program successfully reads the raw bytes at the beginning of the file and cross-references them against known signatures. It correctly identifies the file's true nature, even though the visual file extension suggests otherwise.
